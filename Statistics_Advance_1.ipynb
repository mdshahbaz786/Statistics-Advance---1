 1.   What is a random variable in probability theory?

It assigns a numerical value to each possible outcome in the sample space.
2. What are the types of random variables?
   - (i). Discrete Random Variable

A discrete random variable takes countable values (finite or infinite countable).

(ii) Continuous Random Variable

A continuous random variable takes uncountable infinite values within an interval.
3. Explain the difference between discrete and continuous distributions.
   - A discrete distribution describes outcomes that are countable, while a continuous distribution describes outcomes that are measurable and can take infinitely many values within a range.
   | Feature                           | Discrete Distribution                 | Continuous Distribution                        |
| --------------------------------- | ------------------------------------- | ---------------------------------------------- |
| **Type of Values**                | Countable, distinct values (0, 1, 2…) | Uncountably infinite values within an interval |
| **Probability Function**          | Probability Mass Function (PMF)       | Probability Density Function (PDF)             |
| **Probability of a Single Value** | P(X = x) can be > 0                   | P(X = x) = 0 always                            |
| **Examples**                      | Binomial, Poisson                     | Normal, Exponential                            |
| **Nature of Data**                | Counting data (number of events)      | Measurement data (height, weight, time)        |

4. What is a binomial distribution, and how is it used in probability?
   - A binomial distribution describes the probability of getting a certain number of successes in a fixed number of independent trials, where each trial has only two outcomes (success or failure) and the probability of success (p) remains constant.

Uses:

- Calculating probability of getting k successes in n trials

- Coin toss problems
- Quality control (defective vs non-defective items)
- Customer purchase probability

- Medical test predictions
## 5.  What is the standard normal distribution, and why is it important?
   -The standard normal distribution is a special type of normal distribution with:

- Mean = 0
- Standard deviation = 1
- Bell-shaped, symmetric curve

 It is represented using the Z-score, which tells how many standard deviations a value is from the mean.

 The standard normal distribution is important because:

(i) It allows us to convert any normal distribution into a standard form using Z-scores.

(ii)It is used for probability calculations and statistical tables (Z-table).

(iii) It forms the basis of confidence intervals, hypothesis testing, and many other statistical methods.

(iv) It helps simplify real-world data analysis by standardizing different datasets.
6. What is the Central Limit Theorem (CLT), and why is it critical in statistics?
   -The Central Limit Theorem (CLT) states that when we take repeated random samples from any population, the distribution of the sample means becomes approximately normal as the sample size becomes large (usually n ≥ 30), even if the original population is not normal.

Why CLT is critical in statistics:

- Allows normal approximation even when population distribution is unknown.
- Enables confidence intervals for estimating population means.
- Forms the foundation of hypothesis testing.
- Simplifies analysis because normal distribution is easy to work with.
- Helps in making reliable predictions about population parameters based on sample data.
7. What is the significance of confidence intervals in statistical analysis?
   - A confidence interval gives a range of values within which the true population parameter is likely to lie.
It shows how accurate and reliable a sample estimate is.
Significance:

- Indicates uncertainty in estimation
- More informative than a single value
- Helps in better decision-making
- Commonly used in research and data analysis
8.  What is the concept of expected value in a probability distribution?
   - The expected value (or mean) of a probability distribution is the weighted average of all possible outcomes, where each value is multiplied by its probability.

Formula (Discrete):

E(X)=∑x⋅P(x)


```python
#Q9. Write a Python program to generate 1000 random numbers from a normal
# distribution with mean = 50 and standard deviation = 5. Compute its mean and standard
# deviation using NumPy, and draw a histogram to visualize the distribution.

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 5, 1000)
mean_val = np.mean(data)
std_val = np.std(data)

print("Mean:", mean_val)
print("Standard Deviation:", std_val)

# Plot histogram
plt.hist(data, bins=30, color = 'cyan')
plt.title("Normal Distribution (Mean = 50, SD = 5)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# You are working as a data analyst for a retail company. The company has
#collected daily sales data for 2 years and wants you to identify the overall sales trend.
#daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
#235, 260, 245, 250, 225, 270, 265, 255, 250, 260]
# ● Explain how you would apply the Central Limit Theorem to estimate the average sales
# with a 95% confidence interval.
# ● Write the Python code to compute the mean sales and its confidence interval.
import numpy as np
from math import sqrt

daily_sales = [
    220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
    235, 260, 245, 250, 225, 270, 265, 255, 250, 260
]

data = np.array(daily_sales)
mean_sales = np.mean(data)
std_sales = np.std(data, ddof=1)

n = len(data)
SE = std_sales / sqrt(n)

lower = mean_sales - 1.96 * SE
upper = mean_sales + 1.96 * SE

print("Mean Sales:", mean_sales)
print("95% Confidence Interval:", (lower, upper))

```
